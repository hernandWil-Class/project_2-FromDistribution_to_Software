# Synthetic follow-up data

All records in this notebook are synthetic. Each row represents follow-up from diagnosis until death or independent right censoring. A censored duration contributes known person-time but is not an event at that time. The observed data intentionally contains no latent event times.

```python
import matplotlib.pyplot as plt

from survivallab.data import fixed_fixture

observations = fixed_fixture()
durations = [row.duration_months for row in observations]
events = [row.event_observed for row in observations]
print(
    f"n={len(observations)}, events={sum(events)}, person-time={sum(durations)} months"
)
```

```python
colors = ["tab:red" if event else "tab:blue" for event in events]
plt.figure(figsize=(8, 4))
plt.scatter(durations, range(1, len(durations) + 1), c=colors)
plt.xlabel("Observed follow-up (months)")
plt.ylabel("Synthetic subject index")
plt.title("Red = event; blue = right-censored")
plt.grid(alpha=0.25)
plt.show()
```

The generator stores the observed minimum of latent event and censor times in `observed.csv`; latent times belong only in `simulation_truth.csv`. Try changing the censoring range while keeping the event-rate parameter fixed: the observed event fraction can change without changing the simulated event-time distribution.
